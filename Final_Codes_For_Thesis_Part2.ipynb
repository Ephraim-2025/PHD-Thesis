# In this section, we will work on the factorization over the $\mathbb{F}_{61}$. We first define our polynomial ring and elliptic curve.

```python
P.<x>=PolynomialRing(GF(61))
E3 = EllipticCurve(GF(61), [20,21])
Curve=E3.points()
```

# In equation 4.4.17 we deduce the polynomial $(20x^3+36x^2+35x+40)-20(x+7)^2$. Let us factorize.

```python
f1=(20*x^3+36*x^2+35*x+40)-20*(x+7)^2
factor(f1)
```

# Now we need to find the corresponding $y$ coordinate for $x=-12$. Note that $x=-12\equiv 49\pmod{61}$

```python
x1=49
Lx=[x1]
y_cord = [t for t in Curve if t[0] in Lx]
y_cord
```
